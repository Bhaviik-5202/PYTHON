### 04) WAP to find the longest word(s) in a file.

```python
fp = open("abc.txt", "r")
data = fp.read()
words = data.split()
max_len = 0
longest_words = []

for w in words:
    if len(w) > max_len:
        max_len = len(w)
        longest_words = [w]
    elif len(w) == max_len:
        longest_words.append(w)

print("Longest word length:", max_len)
print("Longest word(s):", longest_words)
fp.close()
```

### 05) WAP to count the no. of lines, words and characters in a given text file.

```python
fp = open("abc.txt", "r")
lines = fp.readlines()
line_count = len(lines)
word_count = 0
char_count = 0

for line in lines:
    words = line.split()
    word_count += len(words)
    char_count += len(line)

print("Number of Lines:", line_count)
print("Number of Words:", word_count)
print("Number of Characters:", char_count)
fp.close()
```

### 08) WAP to create an UDF named frequency to count occurances of the specific word in a given text file.

```python
fp = open("abc.txt", "r")
search = input("Enter word to search: ")
count = 0
for line in fp:
    words = line.split()
    count += words.count(search)
print("Occurrence of", search, "=", count)
fp.close()
```

### 09) WAP to get the score of five subjects from the user, store them in a file. Fetch those marks and find the highest score.

```python
fp = open("marks.txt", "w")
for i in range(5):
    m = int(input("Enter marks: "))
    fp.write(str(m) + "\n")
fp.close()

fp = open("marks.txt", "r")
highest = 0
for line in fp:
    mark = int(line.strip())
    if mark > highest:
        highest = mark
print("Highest Score =", highest)
fp.close()
```

### 10) WAP to write first 100 prime numbers to a file named primenumbers.txt 
(Note: each number should be in new line)

```python
fp = open("primenumbers.txt", "w")
count = 0
num = 2
while count < 100:
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        fp.write(str(num) + "\n")
        count += 1
    num += 1
fp.close()
print("First 100 prime numbers written to primenumbers.txt")
```

### 13) Demonstrate tell() and seek() for all the cases(seek from beginning-end-current position) taking a suitable example of your choice.

```python
fp = open("abc.txt", "rb") # binary mode

print("Initial position:", fp.tell())

# 1. From BEGINNING
fp.seek(6, 0)
print("\nAfter seek from beginning:", fp.tell())
print("Data:", fp.read(6).decode())

# 2. From CURRENT position
fp.seek(1, 1)
print("\nAfter seek from current:", fp.tell())
print("Data:", fp.read(6).decode())

# 3. From END
fp.seek(-5, 2)
print("\nAfter seek from end:", fp.tell())
print("Data:", fp.read().decode())

fp.close()
```
