### 01) Write a function to calculate BMI given mass and height. (BMI = mass/h**2)

```python
def calculate_BMI(mass, height):
    bmi = mass / (height ** 2)
    return bmi

weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

result = calculate_BMI(weight, height)
print("Your BMI is:", round(result, 2))
```

### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def is_prime(n):
    if n <= 1:
        return 0
    for i in range(2, n):
        if n % i == 0:
            return 0

    return 1

num = int(input("Enter number: "))
print(is_prime(num))
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
def primes_range(a, b):
    prime_list = []
    for num in range(a, b + 1):
        if num > 1:
            for i in range(2, num):
                if num % i == 0:
                    break
            else:
                prime_list.append(num)
    return prime_list

start = int(input("Enter start number: "))
end = int(input("Enter end number: "))
print(primes_range(start, end))
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)


n = int(input("Enter N: "))

print("Fibonacci Series:")
for i in range(n):
    print(fibonacci(i), end=" ")
    
print("\nNth Fibonacci term is:", fibonacci(n-1))
```

### 09) Write a function to get the name of the student based on the given rollno.
#### Example: Given dict1 = {101:'Ajay', 102:'Rahul', 103:'Jay', 104:'Pooja'} find name of student whose rollno = 103

```python
def get_student_name(dict1, rollno):
    for k in dict1:
        if k == rollno:
            return dict1[k]
    return "Roll number not found"

dict1 = {101:'Ajay', 102:'Rahul', 103:'Jay', 104:'Pooja'}
roll = int(input("Enter roll number: "))
print(get_student_name(dict1, roll))
```

### 10) Write a function to get the sum of the scores ending with zero.
#### Example : scores = [200, 456, 300, 100, 234, 678] 
#### Ans = 200 + 300 + 100 = 600

```python
# def sum_ending_zero(scores):
#     return sum(s for s in scores if s % 10 == 0)

def sum_ending_zero(scores):
    total = 0
    for s in scores:
        if s % 10 == 0:
            total += s
    return total

scores = [200, 456, 300, 100, 234, 678]
print(sum_ending_zero(scores))
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
# def invert_dict(d):
#     return {v: k for k, v in d.items()}

def invert_dict(d):
    new_dict = {}
    for k, v in d.items():
        new_dict[v] = k
    return new_dict

dict1 = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
print(invert_dict(dict1))
```

### 12) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
def count_upper_lower(s):
    upper = 0
    lower = 0
    for ch in s:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1
    return upper, lower

s1 = input("Enter string: ")
u, l = count_upper_lower(s1)
print("No of Uppercase =", u)
print("No of Lowercase =", l)
```

### 14) For the given list of names of students, extract the names having more that 7 characters. Use filter().

```python
students = ["Rahul", "Alexander", "Priyanka", "Jay", "Samantha", "Amitabh"]

long_names = list(filter(lambda name: len(name) > 7, students))

print(long_names)
```

### 15) For the given list of names of students, convert the first letter of all the names into uppercase. use map().

```python
students = ["rahul", "alexander", "priya", "jay", "samantha"]

capital_names = list(map(lambda name: name.capitalize(), students))

print(capital_names)
```

### 16) Write udfs to call the functions with following types of arguments:
1. Positional Arguments
2. Keyword Arguments
3. Default Arguments
4. Variable Legngth Positional(*args) & variable length Keyword Arguments (**kwargs)
5. Keyword-Only & Positional Only Arguments

```python
# 1. Positional Arguments
def add(a, b):
    print("Sum =", a + b)

add(10, 20)
```

```python
# 2. Keyword Arguments
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(age=20, name="Rahul")
```

```python
# 3. Default Arguments
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Ajay")
```

```python
# 4. Variable Length Arguments
# (a) *args — Variable Positional
def total(*nums):
    print("Sum =", sum(nums))

total(10, 20, 30)

# (b) **kwargs — Variable Keyword
def details(**info):
    for k, v in info.items():
        print(k, ":", v)

details(name="Pooja", city="Ahmedabad", age=21)
```

```python
# 5. Positional-Only & Keyword-Only Arguments
def demo(a, b, /, c, d, *, e, f):
    print(a, b, c, d, e, f)

demo(1, 2, 3, d=4, e=5, f=6)
```
