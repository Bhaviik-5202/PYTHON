### Practical 1: Count Lines, Words and Characters

**Problem Definition:**  
Write a Python program to read text from a file and count the total number of lines, words, and characters. The result should be written to an output file.


```python
fp1 = open("OutputFiles/abc.txt", 'r')
text = fp1.read()
fp1.close()

lines_count = text.count('\n') + 1
words_count = len(text.split())
char_count = len(text)

fp2 = open("OutputFiles/myfile.txt", 'w')
fp2.write("Total Lines : " + str(lines_count) + "\n")
fp2.write("Total Words : " + str(words_count) + '\n')
fp2.write("Total Char : " + str(words_count) + '\n')
fp2.close()
```


### Practical 3: Student Marks and Average

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Store marks in a list, calculate the average using a user-defined function, and write student details along with average marks to an output file.


```python
def average(m1, m2, m3):
    return (m1 + m2 + m3) / 3

fin = open("OutputFiles/avgdata.txt", 'r')
fout = open("OutputFiles/result.txt", 'w')

for line in fin:
    roll, name, m1, m2, m3 = line.split()
    
    m1 = int(m1)
    m2 = int(m2)
    m3 = int(m3)
    
    avg = average(m1, m2, m3)
    
    fout.write(roll + " " + name + " Average: " + str(avg) + "\n")

fin.close()
fout.close()
```


### Practical 4: Maximum and Minimum using Tuple

**Problem Definition:**  
Write a Python program to read numbers from a file, store them in a tuple, and find the maximum and minimum values. Write the result to an output file.


```python
fin = open("OutputFiles/numbers.txt", "r")
data = fin.read().split()
nums = tuple(map(int,data))
fin.close()

fout = open("OutputFiles/output.txt", "w")
fout.write("Maximum: " + str(max(nums)) + "\n")
fout.write("Minimum: " + str(min(nums)) + "\n")
fout.close()
```


### Practical 5: Remove Duplicate Words using Set

**Problem Definition:**  
Write a Python program to read words from a file and remove duplicate words using a set. Store the unique words in an output file.


```python
fin = open("OutputFiles/abc.txt", "r")
words = set(fin.read().split())
fin.close()

fout = open("OutputFiles/rabc.txt", "w")
for w in words:
    fout.write(w + " ")
fout.close()
```


### Practical 6: Word Frequency using Dictionary

**Problem Definition:**  
Write a Python program to count the frequency of each word in a text file using a dictionary and write the result to an output file.


```python
with open("OutputFiles/abc.txt", "r") as fin:
    data = fin.read().lower()

words = data.split()

freq = {}
for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

with open("OutputFiles/output3.txt", "w") as fout:
    for key in freq:
        fout.write(key + " : " + str(freq[key]) + "\n")
```


### Practical 7: Student Result and Grade Calculation

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Use a user-defined function to determine grade (A/B/C/Fail) and write the student result to an output file.

**Grade Criteria:**  
- Marks ≥ 75 : Grade A  
- Marks ≥ 60 : Grade B  
- Marks ≥ 40 : Grade C  
- Marks < 40 : Fail


```python
def grade(m):
    if m >= 75:
        return "A"
    elif m >= 60:
        return "B"
    elif m >= 40:
        return "C"
    else:
        return "Fail"

fin = open("OutputFiles/data2.txt", "r")
fout = open("OutputFiles/finalres.txt", "w")

for line in fin:
    roll, name, mark = line.split()
    fout.write(roll + " " + name + " Grade: " + grade(int(mark)) + "\n")

fin.close()
fout.close()
```
