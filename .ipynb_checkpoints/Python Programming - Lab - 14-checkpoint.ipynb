### 04) WAP to plot a line chart showing temperature variation during a week. Mark the lowest and highest temperatue.

```python
Days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
Temperature (°C) = [20, 32, 31, 33, 34, 43, 36]
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temp = [20, 32, 31, 33, 34, 43, 36]

min_temp = min(temp)
max_temp = max(temp)

min_index = temp.index(min_temp)
max_index = temp.index(max_temp)

# Plot line chart
plt.plot(days, temp, marker='o', label="Temperature")

# Highlight min and max points
plt.scatter(days[min_index], min_temp, label="Lowest Temp")
plt.scatter(days[max_index], max_temp, label="Highest Temp")

# Annotate points
plt.text(days[min_index], min_temp, f"{min_temp}°C")
plt.text(days[max_index], max_temp, f"{max_temp}°C")

# Labels and title
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Variation")

plt.legend()
plt.grid()

plt.show()
```

### 05) WAP to create a detailed pie chart representing the monthly expenditure distribution of a family. 
### The chart should display labels, percentage values, title, different colors for each slice, and explode one slice to highlight the highest expense.

```python
Categories = ["Rent", "Food", "Education", "Transport", "Entertainment", "Savings"]
Amount (₹) = [15000, 8000, 6000, 3000, 2000, 5000]
```

```python
import matplotlib.pyplot as plt
Categories = ["Rent", "Food", "Education", "Transport", "Entertainment", "Savings"]
Amount = [15000, 8000, 6000, 3000, 2000, 5000]
c = ["#C71585", "#4682B4", "#32CD32", "#FFD700", "#9370DB", "#8B4513"]

plt.pie(Amount, labels = Categories, autopct=".%1.2f%%", colors=c, explode=[0,0,0.2,0,0,0], radius=1, shadow=True)
plt.title("")
plt.show()


# import matplotlib.pyplot as plt

# categories = ["Rent", "Food", "Education", "Transport", "Entertainment", "Savings"]
# amount = [15000, 8000, 6000, 3000, 2000, 5000]

# max_index = amount.index(max(amount))

# # Create explode list (all 0)
# explode = [0, 0, 0, 0, 0, 0]

# explode[max_index] = 0.1

# # Plot pie chart
# plt.pie(amount,
#         labels=categories,
#         autopct='%1.1f%%',
#         explode=explode)

# plt.title("Monthly Expenditure")
# plt.show()
```

### 06) WAP to create bar chart showing the marks obtained by students in different subjects.
### The program should also display the marks on the top of the respective bar.

```python
Subjects = ["Python", "DBMS", "SE", "PS", "PE-1"]
Marks = [78, 85, 72, 80, 90]
```

```python
# import matplotlib.pyplot as plt
# Subjects = ["Python", "DBMS", "SE", "PS", "PE-1"]
# Marks = [78, 85, 72, 80, 90]
# bars = plt.bar(Subjects, Marks)
# plt.bar_label(bars)
# plt.xlabel("Subjects")
# plt.ylabel("Marks")
# plt.title("Student's Performance of Sem 4")
# plt.show()


import matplotlib.pyplot as plt

subjects = ["Python", "DBMS", "SE", "PS", "PE-1"]
marks = [78, 85, 72, 80, 90]

plt.bar(subjects, marks)

for i in range(len(subjects)):
    plt.text(i, marks[i], marks[i], ha='center', va='bottom')

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Marks Chart")
plt.show()
```

### 07) WAP to create a horizontal bar chart representing the number of books sold in different categories. 
### The program should also display the number of books sold at the end of the respective bar.

```python
Categories = ["Fiction", "Science", "History", "Technology", "Biography"]
Books Sold = [120, 90, 70, 110, 60]
```

```python
# import matplotlib.pyplot as plt
# Categories = ["Fiction", "Science", "History", "Technology", "Biography"]
# BooksSold = [120, 90, 70, 110, 60]
# bars = plt.barh(Categories, BooksSold)
# plt.bar_label(bars)
# plt.xlabel("Categories")
# plt.ylabel("Books Sold")
# plt.title("Student's Performance of Sem 4")
# plt.show()

import matplotlib.pyplot as plt

categories = ["Fiction", "Science", "History", "Technology", "Biography"]
books = [120, 90, 70, 110, 60]

plt.barh(categories, books)

# Add values using loop
for i in range(len(categories)):
    plt.text(books[i], i, books[i], va='center')

plt.xlabel("Books Sold")
plt.title("Horizontal Bar Chart")

plt.show()
```

### 08) WAP to create a scatter plot representing the relationship between advertising expenditure and sales revenue of a company.

```python
Advertising Cost (₹ thousands) = [5, 8, 10, 12, 15, 18, 20]
Sales Revenue (₹ thousands) = [50, 60, 65, 70, 80, 90, 95]
```

```python
import matplotlib.pyplot as plt

ad_cost = [5, 8, 10, 12, 15, 18, 20]
sales = [50, 60, 65, 70, 80, 90, 95]

plt.scatter(ad_cost, sales)

plt.xlabel("Advertising Cost (₹ thousands)")
plt.ylabel("Sales Revenue (₹ thousands)")
plt.title("Advertising vs Sales Relationship")

plt.grid()
plt.show()

# import matplotlib.pyplot as plt
# AdvertisingCost = [5, 8, 10, 12, 15, 18, 20]
# SalesRevenue = [50, 60, 65, 70, 80, 90, 95]
# city = ['Rajkot', 'Morbi', 'Jamnagar', 'Dwarka', 'Surat', 'Ahm', 'Gondal']

# plt.scatter(AdvertisingCost, SalesRevenue, c=AdvertisingCost , cmap="rainbow")
# plt.xlabel("Advertising Expenditure")
# plt.ylabel("Sales Revenue")
# plt.colorbar(label='City')
# plt.show()
```

### 09) WAP to create a histogram representing the distribution of marks obtained by students in a class.

```python
Marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 82, 85, 88, 90, 92, 95]
```

```python
# import matplotlib.pyplot as plt
# Marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 82, 85, 88, 90, 92, 95]
# plt.hist(Marks, bins=12, histtype='stepfilled', align='mid', label='Marks', edgecolor='k')
# plt.legend()
# plt.show()


import matplotlib.pyplot as plt
marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 82, 85, 88, 90, 92, 95]

plt.hist(marks, bins=6, edgecolor='black')

plt.xlabel("Marks Range")
plt.ylabel("Frequency")
plt.title("Student Marks Histogram")

plt.grid(axis='y')
plt.show()
```

### 10) WAP to create a box plot representing the distribution of daily commuting time (in minutes) of employees and identify outliers (if any).

```python
Commuting Time (minutes) = [5, 18, 20, 22, 25, 27, 28, 30, 32, 35, 38, 40, 42, 90]
```

```python
import matplotlib.pyplot as plt

time = [5, 18, 20, 22, 25, 27, 28, 30, 32, 35, 38, 40, 42, 90]

plt.boxplot(time, showfliers=True)

plt.ylabel("Commuting Time (minutes)")
plt.title("Daily Commuting Time Distribution")

plt.show()
```
