### 01) WAP to create Calculator module which defines functions like add, sub,mul and div. 
### Create another .py file that uses the functions available in Calculator module.

```python
import calculator

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Addition:", calculator.add(a, b))
print("Subtraction:", calculator.sub(a, b))
print("Multiplication:", calculator.mul(a, b))
print("Division:", calculator.div(a, b))
```

### 02) WAP to pick a random character from a given String.

```python
import random
s = input("Enter a string: ")
index = random.randint(0, len(s)-1)
print("Random character:", s[index])

# OR
import random
s = input("Enter a string: ")
ch = random.choice(s)
print("Random character:", ch)
```

### 03) WAP to pick a random element from a given list.

```python
import random
my_list = [10, 20, 30, 40, 50]
index = random.randint(0, len(my_list)-1)
print("Random element:", my_list[index])

# OR
import random
my_list = [10, 20, 30, 40, 50]
element = random.choice(my_list)
print("Random element from list:", element)
```

### 04) WAP to roll a dice in such a way that every time you get the same number.

```python
import random
random.seed(1)  # Set seed value
dice = random.randint(1, 6)
print("Dice number:", dice)

# OR
import random
random.seed(5)
print("Dice number:", random.randint(1,6))
```

### 05) WAP to generate 3 random integers between 100 and 999 which is divisible by 5.

```python
import random
for i in range(3):
    num = random.randrange(100,1000,5)
    print(num)


# OR
import random
count = 0
while count < 3:
    num = random.randint(100, 999)

    if num % 5 == 0:
        print(num)
        count += 1
```

### 06) WAP to generate 100 random lottery tickets and pick two lucky tickets from it and announce them as Winner and Runner up respectively.

```python
import random
l1 = [random.randrange(100,999) for i in range(100)]
lucky = random.sample(l1, k = 2)
print("Winner Ticket : ", lucky[0])
print("Runner Up : ", lucky[1])

#OR
import random
tickets = random.sample(range(100000,999999),100)
winner, runner_up = random.sample(tickets,2)
print("Winner:", winner)
print("Runner Up:", runner_up)
```

### 07) WAP to print current date and time in Python.

```python
import datetime
now = datetime.datetime.now()
print("Current Date and Time:", now)

#OR
import datetime
print(datetime.datetime.now())

#OR
import datetime
now = datetime.datetime.now()
print("Date:", now.date())
print("Time:", now.time())
```

### 08) Subtract a week (7 days) from a given date in Python.

```python
import datetime
dt = datetime.datetime.now() - datetime.timedelta(days = 7)
print(dt)
```

### 09) WAP to Calculate number of days between two given dates.

```python
import datetime
d1 = datetime.date(2025,1,1)
d2 = datetime.date(2022,1,1)
dif = d2 - d1
print(dif.days)

#OR
from datetime import datetime, timedelta
date_str = input("Enter date (YYYY-MM-DD): ")
d = datetime.strptime(date_str, "%Y-%m-%d")
new_date = d - timedelta(days=7)
print("Date after subtracting 7 days:", new_date.date())


#OR
from datetime import datetime, timedelta
d = datetime(2026,3,11)
print(d - timedelta(days=7))
```

### 10) WAP to Find the day of the week of a given date.(i.e. wether it is sunday/monday/tuesday/etc.)

```python
import datetime
date = datetime.datetime.now()
print(date.strftime("%A"))

#OR
from datetime import datetime
d = datetime(2026,3,15)
print(d.strftime("%A"))
```
